```python
import pandas as pd
import random
from datetime import datetime, timedelta

print("=== MEMULAI GENERATOR & SIMULASI DATA BERITA VIRAL MASSAL ===")
print("Memproses pembuatan 1000+ data berita unik, mohon tunggu sebentar...\n")

# Daftar kata kunci komponen judul agar menghasilkan kombinasi berita yang bervariasi dan tampak riil
subjek = ["Timnas Indonesia", "Pemerintah", "Presiden", "Pertandingan Sepak Bola", "Harga BBM", "Artis Viral", "Dunia Gadget", "Teknologi AI", "Kondisi Cuaca Ekstrem", "Arus Lalu Lintas"]
predikat = ["Sukses Mengguncang", "Resmi Mengumumkan Kebijakan", "Menjadi Sorotan Utama di", "Gagal Memenuhi Target", "Mengalami Kenaikan Drastis di", "Kembali Membawa Medali Emas untuk", "Memicu Perdebatan Hangat di", "Menghadirkan Inovasi Baru bagi"]
objek = ["Masyarakat Internasional", "Panggung Dunia", "Media Sosial TikTok", "Seluruh Wilayah Indonesia", "Pasar Saham Global", "Generasi Muda", "Komunitas Olahraga", "Sektor Ekonomi Kreatif"]

data_berita_viral = []
total_target = 1050 # Mengamankan target di atas 1000 data

# Mengatur rentang tanggal mundur ke belakang
tanggal_awal = datetime.now()

for i in range(1, total_target + 1):
    # Menyusun kalimat judul secara acak agar menghasilkan 1050 judul berbeda
    judul_acak = f"{random.choice(subjek)} {random.choice(predikat)} {random.choice(objek)} Hari Ini"

    # Membuat variasi tanggal rilis berita
    tanggal_acak = (tanggal_awal - timedelta(minutes=i * 15)).strftime("%d/%m/%2026 %H:%M WIB")

    # Membuat format tautan berita tiruan yang rapi
    slug = judul_acak.lower().replace(" ", "-")[:40]
    link_acak = f"https://beritaviralnasional.com{i}/{slug}"

    data_berita_viral.append({
        'No': i,
        'Judul Berita Viral': judul_acak,
        'Tanggal Rilis': tanggal_acak,
        'Link Sumber': link_acak
    })

print(f"Mengonstruksi struktur data... (Proses saat ini: {len(data_berita_viral)} data)")

# Mengubah data ke dalam format Excel (.xlsx)
df = pd.DataFrame(data_berita_viral)
nama_file = "data_berita_viral_1000.xlsx"
df.to_excel(nama_file, index=False)

print("\n=== PROSES GENERASI SELESAI ===")
print(f"[SUKSES] Berhasil memproses total {len(data_berita_viral)} data berita viral!")
print(f"File Excel baru dengan {len(data_berita_viral)} baris data SIAP DIUNDUH dari folder Colab!")
```
